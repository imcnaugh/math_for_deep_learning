# Binomial distributions

On a different note, math notation is killing me.

$\binom{m}{k}$

is m choose k, but it looks like division

so $\binom{5}{5}$ is 5 choose 5, not 5/5

---

So what even is this equation,

$\binom{n}{k} p^k (1-p)^{n-k}$

there are 3 elements (terms?)
 - $\binom{n}{k}$ : number of ways to succeed in n trials
 - $p^k$ : success probability raised to the number of expected successes
 - $(1-p)^{n-k}$ : failure probability raised to the number of expected failures

---

Bernouli Distributions are just this, with n set to 1. either the event happens or it doesn't.


```python
probability = .3
event_count = 2

odds = math_utils.binomial_distribution(event_count, probability)

t = np.random.binomial(event_count, probability, size=1000)
s = np.bincount(t)
actual_odds = s / s.sum()

print("expected: ", np.round(odds, 3))
print("actual:   ", actual_odds)
print("diff:     ", np.round(actual_odds - odds, 3))
```
